# Mathematica（Wolfram Language）の例

このノートブックは **Wolfram Language 13.1** カーネルで実行します（Cursor 右上の「Select Kernel」→「Jupyter Kernel...」→「Wolfram Language 13.1」）。

Quarto の原稿（`.qmd`）は1つの文書で1つのカーネルしか使えないため、Mathematica の結果は **図やデータとしてファイルに書き出し**、原稿から読み込みます。

- 図: `manuscript/figures/` に PDF（論文用）と PNG（Web用）で書き出す
- 数値: `manuscript/data/` に CSV で書き出す


```
(* 作業フォルダの確認。notebooks フォルダが表示されれば OK *)
Directory[]
```

```
(* 図を作る *)
plot = Plot[{Sin[x], Cos[x]}, {x, 0, 2 Pi},
  PlotLegends -> {"sin x", "cos x"},
  AxesLabel -> {"x", None},
  ImageSize -> 400]
```

```
(* 原稿から読み込めるように manuscript/figures/ へ書き出す *)
figDir = FileNameJoin[{ParentDirectory[Directory[]], "manuscript", "figures"}];
Quiet[CreateDirectory[figDir]];
Export[FileNameJoin[{figDir, "mma-trig.pdf"}], plot];
Export[FileNameJoin[{figDir, "mma-trig.png"}], plot, ImageResolution -> 200];
FileNames["mma-trig.*", figDir]
```

```
(* 数値結果を CSV で書き出す例 *)
dataDir = FileNameJoin[{ParentDirectory[Directory[]], "manuscript", "data"}];
Quiet[CreateDirectory[dataDir]];
data = Table[{x, N[Sin[x]]}, {x, 0, 2 Pi, Pi/8}];
Export[FileNameJoin[{dataDir, "mma-sin.csv"}], Prepend[data, {"x", "sin_x"}]]
```
